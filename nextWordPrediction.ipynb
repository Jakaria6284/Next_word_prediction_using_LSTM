```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,Dense
```

```python
text="""It was a bright cold day in April, and the clocks were striking thirteen." Big Brother is watching you. War is peace. Freedom is slavery. Ignorance is strength. The past was erased, the erasure was forgotten, the lie became the truth. Doublethink means the power of holding two contradictory beliefs in one's mind simultaneously, and accepting both of them. If you want a picture of the future, imagine a boot stamping on a human face—forever. Nothing was your own except the few cubic centimeters inside your skull. We are the dead. Our only true life is in the future. Thoughtcrime does not entail death: thoughtcrime is death. To die hating them, that was freedom.

Reality exists in the human mind, and nowhere else. Freedom is the freedom to say that two plus two make four. Power is in tearing human minds to pieces and putting them together again in new shapes of your own choosing. The choice for mankind lies between freedom and happiness, and for the great bulk of mankind, happiness is better. Who controls the past controls the future. Who controls the present controls the past. The best books are those that tell you what you know already. Until they become conscious, they will never rebel, and until after they have rebelled, they cannot become conscious. It is not possible to found a civilization on fear and hatred and cruelty. It would never endure. Perhaps one did not want to be loved so much as to be understood. The object of power is power"""
```

```python
tokenizer=Tokenizer()
tokenizer.fit_on_texts([text])
total_words=len(tokenizer.word_index)+1
tokenizer.word_index.items()
```

```python
input_sequences=[]
for line in text.split("."):
  word=tokenizer.texts_to_sequences([line.strip()])[0]
  for i in range(1,len(word)):
    n_gram_sequence=word[:i+1]
    input_sequences.append(n_gram_sequence)


```

```python
max_seq_length = max(len(seq) for seq in input_sequences)
input_sequences = pad_sequences(input_sequences, maxlen=max_seq_length, padding='pre')
input_sequences
```

```python
X,y=input_sequences[:,:-1],input_sequences[:,-1]
print(y)
```

```python
y=tf.keras.utils.to_categorical(y,num_classes=total_words)
print(y)
```

```python
model=Sequential(
   [
    Embedding(total_words,10,input_length=max_seq_length-1),
    LSTM(100,return_sequences=True),
    LSTM(100),
    Dense(100,activation="relu"),
    Dense(total_words,activation="softmax")
   ]
)
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
model.fit(X,y,epochs=500,verbose=1)
```

```python
loss, accuracy = model.evaluate(X, y)
print(f"Test Accuracy: {accuracy * 100:.2f}%")
```

```python
def predict_next_word(model, tokenizer, text_seq, max_length):
    encoded = tokenizer.texts_to_sequences([text_seq])[0]
    encoded = pad_sequences([encoded], maxlen=max_length-1, padding='pre')
    predicted_index = np.argmax(model.predict(encoded), axis=-1)[0]

    for word, index in tokenizer.word_index.items():
        if index == predicted_index:
            return word
    return ""

# Test next word prediction
seed_text = "Reality exists in the"
predicted_word = predict_next_word(model, tokenizer, seed_text, max_seq_length)
print(f"Predicted next word: {predicted_word}")
```
